```python
import pandas as pd
df = pd.read_csv('../local_data/silver/EIA_pre_feature.csv', parse_dates=["timestamp"])
```

```python
cols = [
    'demand_forecast_mwh',
    'actual_demand_mwh',
    'net_generation_mwh',
    'net_interchange_mwh'
]

mask = df[cols].isna().any(axis=1)
groups = (mask != mask.shift()).cumsum()

gap_details = []

for g, group in df[mask].groupby(groups):
    start = group.index[0]
    end = group.index[-1]
    length = len(group)
    
    missing_counts = group[cols].isna().sum()
    
    gap_details.append({
        'start': start,
        'end': end,
        'length_hours': length,
        'missing_counts': missing_counts.to_dict()
    })

gap_details
```
